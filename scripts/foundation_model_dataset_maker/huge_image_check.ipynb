```python
import os
import glob
from PIL import Image
from tqdm import tqdm

# Ton dossier d'images
DATA_DIR = "/home/lmalerba/projects/sdim_data_challenge/kaggle_challenge_sdi_cleaned/kaggle_challenge_sdi/data/pelgas_dataset/101141/individual_images" 
all_files = glob.glob(os.path.join(DATA_DIR, "**", "*.jpg"), recursive=True)

print("Recherche des images géantes...")

for filepath in tqdm(all_files):
    try:
        # On ouvre juste pour lire la taille, sans charger les pixels (lazy loading)
        with Image.open(filepath) as img:
            pixels = img.width * img.height
            # La limite par défaut de Pillow est 89478485
            if pixels > 89000000:
                print(f"!!! GÉANTE DÉTECTÉE !!!")
                print(f"Nom : {filepath}")
                print(f"Taille : {img.width} x {img.height} ({pixels} pixels)")
                print("-" * 20)
    except Exception as e:
        print(f"Erreur sur {filepath}: {e}")
```
